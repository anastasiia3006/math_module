```python
import numpy as np
import matplotlib.pyplot as plt
import scipy
```

```python
low = 1
high = 21
x_values = np.arange(low, high)
x_values
```

```python
probabilities = scipy.stats.randint.pmf(x_values, low, high)
probabilities
```

```python
print(f'X: {x_values}')
print(f'P(X): {probabilities}')
print(f'Sum of probabilities: {np.sum(probabilities):.2f}')
```

Підкидаємо кубик 3 рази PMF - розподіл

```python
n_trials = 3
prob_succes = 0.5
x_values = np.arange(0, n_trials + 1)

probabilities_2 = scipy.stats.binom.pmf(x_values, n_trials, prob_succes)
print(f'X: {x_values}')
print(f'P(X): {probabilities_2}')
print(f'Sum of probabilities: {np.sum(probabilities_2):.2f}')
```

```python
plt.figure(figsize=(7,4))
plt.bar(x_values, probabilities_2, color='lightsalmon', edgecolor='black')
```

Continues Dist (PDF)

```python
a, b = 10, 20
x_range = np.linspace(a - 2, b + 2, 500)
pdf_range = scipy.stats.uniform.pdf(x_range, loc=a, scale = b - a)
print(f'Щільність ймовірності для рівномірного розподілу на відрізку [{a}, {b}]: {pdf_range[250]}')
```

```python
plt.figure(figsize=(8,5))
plt.plot(x_range, pdf_range, color='purple', linewidth=2)
plt.fill_between(x_range, pdf_range, color='purple', alpha=0.2)
plt.title('PDF для рівномірного неперервного розподілу на [{a}, {b}]')
plt.xlabel('x')
plt.ylabel(f'Щільність ймовірності f(x)')
plt.ylim(bottom=0)
plt.xlim(x_range.min(), x_range.max())

x_interval = np.linspace(11, 14, 100)
pdf_interval = scipy.stats.uniform.pdf(x_interval, loc=a, scale = b - a)
plt.fill_between(x_interval, pdf_interval, color='red', alpha=0.5, label='P(11 ≤ X ≤ 14)')
plt.legend()
plt.show()
```
